```python
with open("input01.txt") as file:
    input = file.readlines()
input
```

```python
input_left, input_right = [], []

for row in input:
    input_left.append(int(row[:5]))
    input_right.append(int(row[8:13]))
input_right
```

```python
input_left.sort()
input_right.sort()
sum  = 0
for left_row, right_row in zip(input_left, input_right):
    diff = abs(left_row - right_row)
    sum += diff
sum
```

```python
sum = 0
for num in input_left:
    product = num * input_right.count(num)
    sum += product
sum
```
